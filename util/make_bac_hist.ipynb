```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('1_dlpnoccsd(t)f122023_ccpvtzf12__m062x_ccpvtz.csv')
```

```python
ref_kcal_mol = df['Reference Enthalpy'].values
calc_kcal_mol = df['Calculated Enthalpy'].values
corr_kcal_mol = df['Corrected Enthalpy'].values
```

```python
before = calc_kcal_mol - ref_kcal_mol
after = corr_kcal_mol - ref_kcal_mol
```

```python
N = 32
plt.hist(before, N, label='Before', alpha=0.7)
plt.hist(after, N, label='After', alpha=0.7)
plt.legend()
plt.xlabel('Hf298 Error kcal/mol')
plt.ylabel('Species Count')
ax = plt.gca()
ax.spines[ 'top'].set_visible(False)
ax.spines['right'].set_visible(False)
```
