# Generic Gibbs energy

```python
x = np.linspace(1, 2, 101)
y = 1.0 + 0.5 * np.float_power(x, 2.0)

colors = plt.rcParams['axes.prop_cycle'].by_key()['color']

plt.plot(x, y, linewidth=6.0, color=colors[0])
plt.fill_between(x=x, y1=y, y2=y+0.25, color=colors[0], alpha=0.1)
plt.fill_between(x=x, y1=y-0.25, y2=y, color=colors[0], alpha=0.1)


ax = plt.gca()
# ax.set_yscale('log')


plt.xlabel(r'T (K)', fontsize=24)
plt.ylabel(r'Species Enthalpy', fontsize=24)

ax = plt.gca()
ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)

ax.spines['bottom'].set_linewidth(6)
ax.spines['left'].set_linewidth(6)
plt.xticks([])
plt.yticks([])
plt.xlim((np.float64(0.95), np.float64(2.05)))
plt.ylim((np.float64(1.425), np.float64(3.075)))
```
