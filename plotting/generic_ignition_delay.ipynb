# Generic sensitivity

```python
N = 7
np.random.seed(400)  # N=7
np.random.seed(401)
np.random.seed(404)
values = sorted(np.random.normal(size=N))




width = 0.5
colors = plt.rcParams['axes.prop_cycle'].by_key()['color']


for k in range(len(values)):
#     plt.barh(k + width * (0 - 1.5), values[k], height=0.7*width, align='edge', color=colors[0])
    plt.barh(k, values[k], height=0.5, align='edge', color=colors[2])
        
ax1 = plt.gca()
ax1.spines['right'].set_visible(False)
ax1.spines['top'].set_visible(False)        
ax1.spines['bottom'].set_linewidth(6)
ax1.spines['left'].set_linewidth(6)
plt.xlabel('Sensitivity', fontsize=24)
plt.ylabel('Parameter', fontsize=24)
plt.xticks([])
plt.yticks([])


```

```python
# Flame speed
```

```python
x = np.linspace(0, 1, 101)
y = 15.0 - np.float_power(x - 0.5, 2.0)

colors = plt.rcParams['axes.prop_cycle'].by_key()['color']

plt.plot(x, y, linewidth=6.0, color=colors[1])

ax = plt.gca()
# ax.set_yscale('log')


plt.xlabel(r'Equivalence Ratio $\phi$', fontsize=24)
plt.ylabel(r'$S_L$ (cm/s)', fontsize=24)

ax = plt.gca()
ax.spines[['right', 'top']].set_visible(False)

ax.spines['bottom'].set_linewidth(6)
ax.spines['left'].set_linewidth(6)
plt.xticks([])
plt.yticks([])
```

```python
# Species concentration
```

```python
x = np.linspace(0, 4, 101)
y = 1.0 - np.exp(-(x - 1))

colors = plt.rcParams['axes.prop_cycle'].by_key()['color']

plt.plot(x, y, linewidth=6.0, color=colors[0])

ax = plt.gca()
# ax.set_yscale('log')


plt.xlabel(r'Time (s)', fontsize=24)
plt.ylabel(r'% Species ', fontsize=24)

ax = plt.gca()
ax.spines[['right', 'top']].set_visible(False)

ax.spines['bottom'].set_linewidth(6)
ax.spines['left'].set_linewidth(6)
plt.xticks([])
plt.yticks([])
```
